```python

datos_crudos = data['businesses'] 

# itero listas con for
lista_procesada = []

for rest in datos_crudos:
    # Extraemos la información de forma segura
    info = {
        'Nombre': rest.get('name'),
        'Rating': rest.get('rating'),
        'Reseñas': rest.get('review_count'),
        'Precio': rest.get('price', 'N/A'),
        'Categorías': ', '.join([c['title'] for c in rest.get('categories', [])]),
        'Ciudad': rest.get('location', {}).get('city'),
        'Dirección': ', '.join(rest.get('location', {}).get('display_address', [])),
        'Teléfono': rest.get('display_phone', 'N/A'),
        'Latitud': rest.get('coordinates', {}).get('latitude'),
        'Longitud': rest.get('coordinates', {}).get('longitude')
    }
    lista_procesada.append(info)

# creo el DataFrame con las listas procesadas
df = pd.DataFrame(lista_procesada)

# aplico filtro para que solo queden los de Denver

df_denver = df[df['Ciudad'].str.contains('Denver', case=False, na=False)].copy()

# guardo en archivo CSV
nombre_csv = 'restaurantes_vegetarianos_denver.csv'
df_denver.to_csv(nombre_csv, index=False, encoding='utf-8-sig')

print(f"✅ ¡Listo! Se filtraron {len(df_denver)} restaurantes de Denver.")
print(f"📂 Archivo generado: {nombre_csv}")

# visualizo los primeros 5 para confirmar
df_denver.head()
```
